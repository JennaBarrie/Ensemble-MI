```python
from glob import glob
from pathlib import Path
import pandas as pd

result_files = glob("results/PhysionetMI/5-fold/**/results.xlsx", recursive=True)
latest_result = max(result_files, key=lambda f: Path(f).stat().st_mtime)

print(f"Results file: {latest_result}")
results_df = pd.read_excel(latest_result)
display(results_df)

accuracy_columns = [c for c in results_df.columns if "acc" in c.lower()]
for column in accuracy_columns:
    values = pd.to_numeric(results_df[column], errors="coerce").dropna()
    if len(values):
        print(f"{column}: mean = {values.mean():.2f}, std = {values.std():.2f}")
```
